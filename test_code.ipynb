```python
from pathlib import Path

plate_data = '/Users/owenburroughs/Desktop/Personal_Skaar_Lab/EPS007_Code/Test_Dataset'
plate_name = '001_BacA-1_EPS007_19700101'

ImageXpress_name_regex = r't(?P<timepoint>\d*)_(?P<row>.)(?P<column>\d*)_s(?P<site>\d*)_w(?P<wavelength>\d*)_z(?P<z_stack>\d*).tif$'
ImageXpress_name_pattern = re.compile(ImageXpress_name_regex)

def extract_metadata_raw_name( well_name:str ):
    print(well_name)
    return ImageXpress_name_pattern.search(str(well_name)).groupdict()


#-------------- Standard FOV Name Functions ---------------
std_name_regex = r'(?P<row>.)(?P<column>\d*)t(?P<timepoint>\d*)z(?P<z_stack>\d*)'

def generate_std_name(groupdict):
    return f'{groupdict['row']}{groupdict['column']}t{groupdict['timepoint']}z{groupdict['z_stack']}'



#-------------- Get unique FOVs given a plate name ----------------------
#
# Files from the ImageXpress come in the following pattern:
# t1_A01_s1_w1_z1.tif

def get_FOVs(wildcards={}):
    plate_dir = plate_data + '/' + plate_name
    return {
        generate_std_name(extract_metadata_raw_name(p))
        for p in Path(plate_dir).iterdir()        
        if ImageXpress_name_pattern.search(str(p)) != None
    }

get_FOVs()
```
